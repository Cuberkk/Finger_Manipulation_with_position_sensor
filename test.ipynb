```python
import numpy as np
```

```python
def z_rotation_matrix(theta):
    """
    Create a rotation matrix for a rotation around the z-axis by an angle theta (in radians).
    
    Parameters:
    theta (float): The angle of rotation in radians.
    
    Returns:
    np.ndarray: A 3x3 rotation matrix.
    """
    theta = np.deg2rad(theta)  # Convert angle from degrees to radians if necessary
    cos_theta = np.cos(theta)
    sin_theta = np.sin(theta)
    
    return np.array([[cos_theta, -sin_theta, 0],
                     [sin_theta,  cos_theta, 0],
                     [0,          0,         1]])
```

```python
rot_interforce_force = z_rotation_matrix(-48)
```

```python
rot_interforce_force
```

```python
rot_interobj_interforce = np.array([
    [0.0, 0.0, 1.0],
    [0.0, 1.0, 0.0],
    [-1.0, 0.0, 0.0]
])
```

```python
rot_object_interobj = z_rotation_matrix(120)  # Rotate 120 degrees around the z-axis
```

```python
rot_object_force = rot_object_interobj @ rot_interobj_interforce @ rot_interforce_force
```
